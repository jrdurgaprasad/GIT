```python
from pyspark.sql import SparkSession

spark = (SparkSession
         .builder
         .appName('Load File')
         .getOrCreate()
        )
```

```python
df = spark.read.csv('customers.csv', inferSchema=True, header=True)
print(df.printSchema())
df.show(truncate=False)
```

```python
from pyspark.sql.functions import col
df_result = df.filter(col('purchase_amount') >= 120)
df_result.show()
df_result.write.mode("overwrite").option("header", "true").csv('results.csv')
```

```python
df_result.coalesce(2).write.mode("overwrite").option("header", "true").csv("results")
```
